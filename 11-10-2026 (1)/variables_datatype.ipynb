Operator precedence

BODMAS --> Assignment

```python
result = 2 + 3 * 4
# 3 * 4 --> 12
# 2 + 12
print(result)
```

```python
result = (2 + 3) * 4 # () -> parentheses
# 2 + 3 -> 5
# 5 * 4 --> 20
print(result)
```
