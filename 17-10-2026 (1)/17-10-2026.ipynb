# shallow copy

```python
a = [1,2,3]
print(id(a))
print(a)
b = a.copy()
print(id(b))
print(b)
```

```python
a = [[1,2,3], [4,5,6]]
print(id(a))
print(a)

b = a.copy()
print(id(b))

a[0].append(456)

print(a)
print(b)
```

# Deep copy

```python
import copy
```

```python
a = [[1,2,3], [4,5,6]]
print(id(a))
print(a)

b = copy.deepcopy(a)
print(id(b))

a[0].append(456)

print(a)
print(b)
```
