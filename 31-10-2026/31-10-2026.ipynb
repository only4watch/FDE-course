# Encapsulation

```python
class Car:
    def __init__(self, brand):
        self.brand = brand # public
        self._ac_voltage = 30 # protected
        self.__current_gps_coordinate = "LA 1.23 LO 4.56" # private

    def run_ac(self):
        if self.check_ac():
            print("Starting AC")

    def check_ac(self):
        if self._ac_voltage > 30:
            return False
        else :
            return True

    def tell_brand(self):
        print(self.brand)

    def accessing_gps_coordinate(self, password):
        if password == 'abc123':
            print(self.__current_gps_coordinate)

    def update_coordinate(self):
        # logic only
        self.__current_gps_coordinate = "LOGIC TO UPDATE COORDINATE"
```

```python
car_obj_1 = Car("Toyota")
```

# Another way

```python
class PayPal:

    def process_payment(self, amount):
        return f"Charing {amount} securely via PayPal"

class GPay:

    def process_payment(self, amount):
        return f"Charing {amount} securely via GPay"

class CreditCard:

    def process_payment(self, amount):
        return f"Charing {amount} securely via CreditCard"

payment_registry = {
    "paypal" : PayPal(),
    "gpay" : GPay(),
    "credit_card" : CreditCard()
}
```

```python
payment_registry
```

```python
def process_payment(payment_gateway, amount):
    try : 
        return payment_registry.get(payment_gateway).process_payment(amount) 
    except :
        return f"Payment gateway {payment_gateway} is not supported."
```

```python
process_payment("credit_card", 100)
```
